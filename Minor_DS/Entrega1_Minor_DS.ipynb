###2.3 Funciones

```python
# @title
# Ordenar dataframe
def contarOrdenar(df1, columna):
    conteo = df1[columna].value_counts()#Te cuenta los datos que se repiten por columna
    conteo_ordenado = conteo.sort_index(ascending=True) # Ordena cada longitud de cinta de menor a mayor
    return conteo_ordenado

# Extraer outliers
def extraer_outliers_iqr(df_in, cols):
  filas_outliers = []
  for col in cols:
    q1 = df_in[col].quantile(0.25)
    q3 = df_in[col].quantile(0.75)
    iqr = q3 - q1
    lim_inf = q1 - 1.5 * iqr
    lim_sup = q3 + 1.5 * iqr

    outliersS = df_in[df_in[col]>lim_sup]
    outliersI = df_in[df_in[col]<lim_inf]

    outliers = pd.concat([outliersI, outliersS])
    filas_outliers.append(outliers)

  return filas_outliers
```

#7 Estadística Inferencial Test de hipótesis


Las hipótesis serán:

1.

H_0: En muestras de acero A300, la ocurrencia de la falla Z-S es independiente de tener un espesor de placa de 70 ± 24,7 mm y un Índice de bordes en Y $\ge$ 0,85.

H_1: En muestras de acero A300, la ocurrencia de la falla Z-S depende de tener un espesor de placa de 70 ± 24,7 mm y un Índice de bordes en Y $\ge$ 0,85.

2.

H0: En muestras de acero A400, la ocurrencia de la falla K-S es independiente de tener un espesor de placa de 40 ± 2,2 mm y un Índice de bordes en Y entre 0,35 y 0  mm.

H1: En muestras de acero A400, la ocurrencia de la falla K-S depende de tener un espesor de placa de 40 ± 2,2 mm y un Índice de bordes en Y entre 0,35 y 0  mm.


¿Por qué 70 mm y 40 mm?

Se obtuvieron estos valores al observar los intercuartiles de las fallas de K-S y Z-S, ambas fallas, contenían intercuartiles de 0, es decir: Q1, mediana y Q3 coincidian con los valores 70mm y 40mm.

##7.1 Test de Hipótesis

*Los rangos ± equivalen a las desviaciones estándar de cada variable

*Con nivel de significancia de alpha=0,05*


```python
# @title
ALPHA = 0.05

# Hipótesis 1
#Filtro por tipo de acero
a300 = df_limpio[df_limpio['TypeOfSteel'] == 'A300']
condicion_1 = a300['Steel_Plate_Thickness'].between(70 - 24.7, 70 + 24.7) & (a300['Edges_Y_Index'] >= 0.85)
tabla_1 = pd.crosstab(condicion_1, a300['TypeOfFault'] == 'Z_Scratch')
stat1, p1, gl1, esperado1 = stats.chi2_contingency(tabla_1)
print(f'Hipótesis 1 -> χ² = {stat1:.4f} | p = {p1:.4g} | grados de libertad = {gl1}')
#Supuesto de que todas las celdas de las frecuencias esperadas sean mayores o iguales a 5 (viendo True o False)
print('Frecuencias esperadas:\n', np.round(esperado1, 2))
print('Cumple supuesto (todas >= 5):', (esperado1 >= 5).all()) #Acá me dará todas True
#Se rechaza o no
print('Se rechaza H₀: hay evidencia de asociación.\n' if p1 < ALPHA else
      'No se rechaza H₀: evidencia insuficiente de asociación.\n')

# Hipótesis 2
#Filtro por tipo de acero
a400 = df_limpio[df_limpio['TypeOfSteel'] == 'A400']
condicion_2 = a400['Steel_Plate_Thickness'].between(40 - 2.2, 40 + 2.2) & (a400['Edges_Y_Index'] <= 0.35)
tabla_2 = pd.crosstab(condicion_2, a400['TypeOfFault'] == 'K_Scratch')
stat2, p2, gl2, esperado2 = stats.chi2_contingency(tabla_2)
print(f'Hipótesis 2 -> χ² = {stat2:.4f} | p = {p2:.4g} | grados de libertad = {gl2}')

#Supuesto de que todas las celdas de las frecuencias esperadas sean mayores o iguales a 5
print('Frecuencias esperadas:\n', np.round(esperado2, 2))
print('Cumple supuesto (todas >= 5):', (esperado2 >= 5).all())#Acá me dará todas True
#Se rechaza o no
print('Se rechaza H₀: hay evidencia de asociación.' if p2 < ALPHA else
      'No se rechaza H₀: evidencia insuficiente de asociación.')
```

#8 Conclusión


En la hipótesis 1 se proporciona evidencia estadística sólida de que, en las placas de acero A300, existe una asociación significativa entre presentar un espesor dentro del rango 70 $\pm$ 24,7 mm junto con un Índice de bordes en Y $\ge 0,85$ y la ocurrencia de fallas de tipo Z-Scratch.


Y por el lado de la hipótesis 2 se concluye que existe una asociación estadísticamente significativa en muestras de acero A400 entre tener un espesor cercano a 40 $\pm$ 2,2 mm combinado con un Índice de bordes en Y $\le 0,35$ y la presencia de defectos de tipo K-Scratch.

Por lo que ambas pruebas confirman que la ocurrencia de ambos tipos de falla no son independientes de las propiedades geométricas evaluadas, validando estadísticamente que combinaciones específicas de espesor de placa e índice de bordes permiten caracterizar y diferenciar estos defectos según el tipo de acero.
